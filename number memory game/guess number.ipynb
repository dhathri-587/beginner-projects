```python
import random
import time

print("🧠 NUMBER MEMORY GAME 🧠")
print("-------------------------")

score = 0
level = 1

while True:
    # Generate random number
    number = random.randint(1000, 9999)

    print("\nLevel:", level)
    print("Remember this number:")
    print(number)

    # Give time to memorize
    time.sleep(3)

    # Clear the output area
    from IPython.display import clear_output
    clear_output()

    # Ask the player
    answer = input("Enter the number you remember: ")

    if answer == str(number):
        score += 10
        print("🎉 Correct!")
        print("Score:", score)
        level += 1
    else:
        print("❌ Wrong!")
        print("The correct number was:", number)
        print("🏆 Final Score:", score)
        break

    # Continue?
    choice = input("\nContinue? (yes/no): ").lower()

    if choice != "yes":
        print("\n🎮 Game Over!")
        print("🏆 Final Score:", score)
        break
```
